```python
import pandas as pd

def load_gita_database(file_path):
    """
    Load the Bhagavad Gita database from a CSV file
    """
    try:
        df = pd.read_csv('Bhagwad_Gita2.csv')
        return df
    except FileNotFoundError:
        return f"Error: File not found at {file_path}"
    except Exception as e:
        return f"Error loading file: {str(e)}"

def lookup_verse(df, chapter, verse):
    """
    Look up a specific verse by chapter and verse number
    """
    try:
        # Find the matching verse
        result = df[(df['Chapter'] == chapter) & (df['Verse'] == verse)]
        
        if len(result) == 0:
            return "Verse not found in the database."
        
        # Extract the meanings
        hindi_meaning = result['HinMeaning'].iloc[0]
        english_meaning = result['EngMeaning'].iloc[0]
        
        return {
            'hindi': hindi_meaning,
            'english': english_meaning
        }
    
    except Exception as e:
        return f"An error occurred: {str(e)}"

def main():
    print("Welcome to Bhagavad Gita Verse Lookup")
    print("=====================================")
    
    # Load the database
    file_path = 'Bhagwad_Gita2.csv'
    df = load_gita_database(file_path)
    
    if isinstance(df, str):  # Error occurred during loading
        print(f"\nError: {df}")
        return
    
    print("\nDatabase loaded successfully!")
    
    while True:
        try:
            # Get user input
            chapter = int(input("\nEnter chapter number (or 0 to exit): "))
            if chapter == 0:
                print("\nThank you for using Bhagavad Gita Verse Lookup!")
                break
                
            verse = int(input("Enter verse number: "))
            
            # Look up the verse
            result = lookup_verse(df, chapter, verse)
            
            if isinstance(result, str):
                print(f"\nError: {result}")
            else:
                print("\nHindi Translation:")
                print(result['hindi'])
                print("\nEnglish Translation:")
                print(result['english'])
                
        except ValueError:
            print("\nError: Please enter valid numbers for chapter and verse.")
        
        print("\n" + "="*40)

if __name__ == "__main__":
    main()
```
